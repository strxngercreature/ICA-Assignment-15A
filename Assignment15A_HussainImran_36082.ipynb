Q5: Make a bar chart of service type vs. average fare

```python
import matplotlib.pyplot as plt
```

```python
avg_fare = rhd.groupby("service_type")["fare"].mean()
plt.bar(avg_fare.index, avg_fare.values, color='blue', width=0.5)
plt.xlabel('service_type')
plt.ylabel('avg_fare')
plt.title('Bar Chart of average fare vs service type')
plt.show()
```
